```python
import torch 
import torch.nn as nn
```

```python
logits = torch.tensor([
    [2.0, 1.0, 0.1, -0.1],
    [0.5, 2.5, 0.3, 0.2],
    [-1.0, 0.0, 3.0, 0.5]
])

labels = torch.tensor([0, 1, 2])
```

```python
softmax_probs = torch.softmax(logits, dim = 1)
print(f"Softmax probability:\n{softmax_probs}")
print(f"Sum of each row: {softmax_probs.sum(dim=1)}")
```

```python
correct_probs = softmax_probs[range(3), labels]
print(correct_probs)
```

```python
ce_manual = -torch.log(correct_probs).mean()
print(f"Manual Cross Entropy Loss: {ce_manual:.4f}")
```

```python
ce_loss = nn.CrossEntropyLoss()
ce_pytorch = ce_loss(logits, labels)
print(f"Pytorch Cross Entropy Loss: {ce_pytorch:.4f}")
```

```python
def softmax_manual(z):
    z_stable = z - z.max(dim = -1, keepdim=True).values
    exp_z = torch.exp(z_stable)
    return exp_z/exp_z.sum(dim=-1, keepdim=True)

z = torch.tensor([2.0, 1.0, 0.1])
print(f"Manual softmax: {softmax_manual(z)}")
print(f"PyTorch softmax: {torch.softmax(z, dim=0)}")
```

```python
for T in [0.1, 1.0, 5.0]:
    probs = torch.softmax(z / T, dim=0)
    print(f"T={T:<3} → {probs.numpy().round(4)}")
```
