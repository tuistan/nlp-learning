```python
import torch
import torch.nn as nn
```

```python
class SimpleRNNCell:
    def __init__(self, input_size, hidden_size):
        self.W_xh = torch.randn(hidden_size, input_size) * 0.01
        self.W_hh = torch.randn(hidden_size, hidden_size) * 0.01
        self.b_h = torch.zeros(hidden_size)

    def forward(self, x_t, h_prev):
        h_t = torch.tanh(self.W_hh @ h_prev + self.W_xh @ x_t + self.b_h)
        return h_t

input_size = 10
hidden_size = 20
seq_len = 5

rnn_cell = SimpleRNNCell(input_size, hidden_size)
h = torch.zeros(hidden_size)

inputs = [torch.randn(input_size) for _ in range(seq_len)]
print(f"RNN sequencial handling:")

for t, x_t in enumerate(inputs):
    h = rnn_cell.forward(x_t, h)
    print(f"Step {t}: h shape={h.shape}, h mean = {h.mean():.4f}, h = {h}")
```
