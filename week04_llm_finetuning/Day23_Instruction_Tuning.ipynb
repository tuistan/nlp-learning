```python
from datasets import Dataset

# Alpaca 模板
ALPACA_TEMPLATE = """Below is an instruction that describes a task. Write a response that appropriately completes the request.

### Instruction:
{instruction}

### Response:
{output}"""

ALPACA_TEMPLATE_WITH_INPUT = """Below is an instruction that describes a task, paired with an input that provides further context. Write a response that appropriately completes the request.

### Instruction:
{instruction}

### Input:
{input}

### Response:
{output}"""


def format_alpaca(example):
    """将单条数据格式化为 Alpaca 模板"""
    if example.get("input", "").strip():
        text = ALPACA_TEMPLATE_WITH_INPUT.format(**example)
    else:
        text = ALPACA_TEMPLATE.format(**example)
    return {"text": text}


# 转为 HuggingFace Dataset
dataset = Dataset.from_list(custom_data)
dataset = dataset.map(format_alpaca)

# 查看格式化结果
print(dataset[0]["text"])
print("=" * 60)
print(f"Dataset size: {len(dataset)}")
print(f"Columns: {dataset.column_names}")
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-1.5B-Instruct")
print(tokenizer.chat_template)  # 应该有值了

def format_chatml(example):
    messages = [
        {"role": "system", "content": "You are an expert NLP engineer. Explain concepts clearly and concisely."},
        {"role": "user", "content": example["instruction"] +
         (f"\n\n{example['input']}" if example.get("input", "").strip() else "")},
        {"role": "assistant", "content": example["output"]}
    ]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=False)
    return {"text": text}

dataset_chatml = Dataset.from_list(custom_data)
dataset_chatml = dataset_chatml.map(format_chatml)
print(dataset_chatml[0]["text"])

```

```python
from trl import SFTTrainer
from transformers import TrainingArguments

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = 'unsloth/Qwen2.5-1.5B-bnb-4bit',
    max_seq_length=1024,
    load_in_4bit=True,
)

model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    lora_alpha=16,
    target_modules=['q_proj', 'k_proj', 'v_proj', 'o_proj', 'gate_proj', 'up_proj', 'down_proj'],
    lora_dropout=0.05,
)

trainer = SFTTrainer(
    model=model,
    train_dataset=dataset_chatml,    # 用 ChatML 格式的数据
    args=TrainingArguments(
    output_dir="./day23_instruction_ft",
    max_steps=30,                      # 直接指定步数，替代 num_train_epochs
    per_device_train_batch_size=2,
    gradient_accumulation_steps=1,     # 改成 1，这样每 step 用 2 条
    learning_rate=2e-4,
    lr_scheduler_type="cosine",
    warmup_ratio=0.1,
    bf16=True,
    logging_steps=5,
    seed=42,
    ),
    dataset_text_field="text",
    max_seq_length=1024,
)

# 4. 开始训练
trainer.train()
```

```python
# 切换到推理模式
FastLanguageModel.for_inference(model)

from transformers import AutoTokenizer
_instruct_tok = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-1.5B-Instruct")
tokenizer.chat_template = _instruct_tok.chat_template

def ask(question, max_tokens=256):
    messages = [
        {"role": "system", "content": "You are an expert NLP engineer."},
        {"role": "user", "content": question}
    ]
    prompt = tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    # 关键：告诉 generate 遇到 <|im_end|> 就停
    stop_token_id = tokenizer.encode("<|im_end|>", add_special_tokens=False)

    im_end_id = tokenizer.convert_tokens_to_ids("<|im_end|>")  # 151645

    outputs = model.generate(
        **inputs,
        max_new_tokens=max_tokens,
        temperature=0.7,
        do_sample=True,
        eos_token_id=[tokenizer.eos_token_id, im_end_id],  # 两个都加
    )
    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )
    return response

# 测试：数据集内的问题（应该回答得不错）
print("=== In-domain ===")
print(ask("What is the difference between BERT and GPT?"))
print()

# 测试：数据集外的问题（看泛化能力）
print("=== Out-of-domain ===")
print(ask("What is knowledge distillation?"))
print()

# 测试：对比 base model（重新加载不带 adapter 的模型看差异）
print("=== Format following ===")
print(ask("List 3 advantages of using LoRA over full fine-tuning."))
```
